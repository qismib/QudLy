# Examples

This notebook provides several examples of how to correctly use the library.

At the moment, QudLy allows users to simulate the action of gates on qudits and perform other operations, as well as measure the state itself.



## Initialization

To start working with the library, the first step is to set the dimension of the qudits to be used. This can only be done once and must be completed before creating any states or gates.


```python
import QudLy as q
import numpy as np

q.set_dim(4)
```

It is now possible to initialize one or more qudits. It is recommended to start with single-qudit states, but it is also possible to directly initialize the total state, provided that its dimension is consistent with the dimension previously set: len(array) = dim^n

The create_state function can be used to combine single-qudit states into the total state.
Regardless of the input vector, it will be automatically normalized.



```python
State_1=q.State([0, 1, 0, 0], 0)
State_2=q.State([0, 0, 1, 0], 1)
State_3=q.State([1, 0, 0, 0], 2)

State=q.create_state([State_1, State_2, State_3])

State_total=q.Total_state([0, 0, 1, 0, 0, 0, 1, 1, 0, 0, 0, 2, 0, 1, 0, 0])


print(State)
print(State_total)
```

Once an initial state has been created, gates can be created to act on the qudits.
To create a gate, it is only necessary to specify which qudits it acts on.
It is always possible to display the matrix associated with a gate by printing it.

```python
Gate_1=q.Gate_Z(2)
Gate_2=q.Gate_H(0)

print(Gate_1)
print(Gate_2)
```

Once the state and gates to be used have been defined, the gates can be applied to the state using the `apply_gate` function. Given a state and a gate, this function returns the resulting final state.

```python
Final_state=q.apply_gate(State, Gate_1)
Final_state=q.apply_gate(Final_state, Gate_2)

print(Final_state)
```

For a better understanding, you can print the state vector in tensor form by using reshape as follows:

`State.state.reshape([dim] * n_qudit)`



```python
print(Final_state.state.reshape([4] * 3))
```

The library also includes several operations that are combinations of multiple gates, such as the Quantum Fourier Transform (QFT) and the SWAP operation.



```python
Final_state_2=q.apply_QFT(State)

Final_state_3=q.apply_SWAP(State, 0, 2)

print(Final_state_2)
print(Final_state_3)
```

Finally, the library allows measurements of either individual qudits or the total state.

The `measure` and `measure_single` functions return an element of the state selected according to the probability associated with each element.

For `measure_single`, by setting `collapse=True`, the function also returns the collapsed state.

```python
measure_1=q.measure(Final_state)

measure_2, Collapsed_state=q.measure_single(Final_state, 0, collapse=True)

print(measure_1)
print(measure_2)
print(Collapsed_state)
```

The `measure_prob` and `measure_prob_single` functions are used to explicitly display the probabilities of the total state or of an individual qudit, respectively.

```python
print(q.measure_prob(Final_state_2))
print(q.measure_prob_single(Final_state_2, 0))
```
